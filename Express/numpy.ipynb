```python
class Quest:
    def __init__(self, q, a):
        self.q = q
        self.a = a


que = [
    "1) Speed of sound is?\n a) 500m/s\n b) 300m/s\n c) 342m/s\n d) 800m/s\n",
    "2) Speed of sound is?\n a) 500m/s\n b) 300m/s\n c) 342m/s\n d) 800m/s\n",
    "3) Speed of sound is?\n a) 500m/s\n b) 300m/s\n c) 342m/s\n d) 800m/s\n",
    "4) Speed of sound is?\n a) 500m/s\n b) 300m/s\n c) 342m/s\n d) 800m/s\n",
    "5) Speed of sound is?\n a) 500m/s\n b) 300m/s\n c) 342m/s\n d) 800m/s\n"
]

ans = [
    Quest(que[0], "c"),
    Quest(que[1], "a"),
    Quest(que[2], "b"),
    Quest(que[3], "c"),
    Quest(que[4], "a")
]

score = 0

for i in ans:
    answer = input(i.q)

    if answer == i.a:
        score = score + 1

print("Total score:", score)
```
