```python
# Simple Recursion Problem
# Print numbers from 1 to N using recursion


def print_numbers(n):

    # Base case:
    # When n becomes 0, stop the recursion.
    if n == 0:
        return

    # Recursive call:
    # First call the function for the smaller number.
    print_numbers(n - 1)

    # Print the current number after the recursive call returns.
    print(n)


# Calling the function
print_numbers(5)
```

```python
# ============================================================
# RECURSION - COMPLEX PRACTICE
# ============================================================
# This program uses recursion to:
#
# 1. Find the sum of digits
# 2. Reverse a number
# 3. Check whether a number is a palindrome
#
# Example:
#     12321
#
# Sum of digits = 1 + 2 + 3 + 2 + 1 = 9
# Reverse      = 12321
# Palindrome   = Yes
# ============================================================


# ------------------------------------------------------------
# 1. Sum of digits using recursion
# ------------------------------------------------------------

def sum_of_digits(n):

    # Base case:
    # A single digit does not need further recursion.
    if n < 10:
        return n

    # Get the last digit.
    last_digit = n % 10

    # Remove the last digit.
    remaining_number = n // 10

    # Add the last digit to the sum of remaining digits.
    return last_digit + sum_of_digits(remaining_number)


# ------------------------------------------------------------
# 2. Reverse a number using recursion
# ------------------------------------------------------------

def reverse_number(n, reverse=0):

    # Base case:
    # When there are no digits left, return the reversed number.
    if n == 0:
        return reverse

    # Extract the last digit.
    last_digit = n % 10

    # Add the digit to the reversed number.
    reverse = reverse * 10 + last_digit

    # Remove the last digit and continue recursion.
    return reverse_number(n // 10, reverse)


# ------------------------------------------------------------
# 3. Check whether a number is a palindrome
# ------------------------------------------------------------

def is_palindrome(n):

    # Store the original number.
    original_number = n

    # Reverse the number using our recursive function.
    reversed_number = reverse_number(n)

    # Compare original and reversed numbers.
    if original_number == reversed_number:
        return True

    return False


# ============================================================
# MAIN PROGRAM
# ============================================================

number = 12321

print("Original Number:", number)

print("Sum of Digits:", sum_of_digits(number))

print("Reversed Number:", reverse_number(number))

if is_palindrome(number):
    print("Palindrome: Yes")
else:
    print("Palindrome: No")


# ============================================================
# EXAMPLE EXECUTION
# ============================================================
#
# sum_of_digits(12321)
#
# 1 + sum_of_digits(1232)
#       ↓
# 2 + sum_of_digits(123)
#       ↓
# 3 + sum_of_digits(12)
#       ↓
# 2 + sum_of_digits(1)
#       ↓
# 1
#
# Returning:
#
# 1
# ↑
# 2 + 1 = 3
# ↑
# 3 + 3 = 6
# ↑
# 2 + 6 = 8
# ↑
# 1 + 8 = 9
#
# Final Answer = 9
# ============================================================
```
