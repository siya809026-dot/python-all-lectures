```python
# Print numbers from 1 to n using recursion

def print_numbers(n):
    if n == 0:
        return

    print_numbers(n - 1)
    print(n)


num = int(input("Enter a number: "))

print("Numbers from 1 to", num, ":")
print_numbers(num)
```

```python
def reverse_number(n, result=0):
    if n == 0:
        return result

    digit = n % 10
    result = result * 10 + digit

    return reverse_number(n // 10, result)


num = int(input("Enter a number: "))

answer = reverse_number(num)

print("Reversed number:", answer)
```

```python
def sum_of_digits(n):
    if n == 0:
        return 0

    digit = n % 10

    return digit + sum_of_digits(n // 10)


num = int(input("Enter a number: "))

answer = sum_of_digits(num)

print("Sum of digits:", answer)
```

```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


def fibonacci_sum(n):
    if n == 0:
        return 0

    return fibonacci(n) + fibonacci_sum(n - 1)


n = 5

print("Sum:", fibonacci_sum(n))
```

```python
def check_fibonacci(n, a=0, b=1):
    if a == n:
        return True

    if a > n:
        return False

    return check_fibonacci(n, b, a + b)


num = 21

if check_fibonacci(num):
    print("Fibonacci number")
else:
    print("Not a Fibonacci number")
```

```python
# Find the sum of digits of a number using recursion

def sum_of_digits(n):
    # Base case
    if n == 0:
        return 0

    # Recursive case
    return (n % 10) + sum_of_digits(n // 10)


# Taking input from user
num = int(input("Enter a number: "))

# Calling the function
result = sum_of_digits(num)

print("Sum of digits:", result)
```

```python
# Reverse a number using recursion

def reverse_number(n, rev=0):
    # Base case
    if n == 0:
        return rev

    # Take last digit
    digit = n % 10

    # Add digit to reversed number
    rev = rev * 10 + digit

    # Recursive call
    return reverse_number(n // 10, rev)


num = int(input("Enter a number: "))

result = reverse_number(num)

print("Reversed number:", result)
```
