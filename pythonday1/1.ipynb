```python
def printNos(n):
    if n == 0:
        
        # base condition
        return
    
    # recursive call first
    printNos(n - 1)
  
    # print after recursion
    print(n, end=' ')


if __name__ == "__main__":
    n = 3
    printNos(n)


#     Step 1

# printNos(3) call hua.

# n == 0? ❌
# printNos(2) call karega
# abhi 3 print nahi hoga
```
